## `package_parse.ipynb`

This basic notebook is used to scrape the project's `.py` and `.ipynb` files for all `import` calls, so as to list them in `/requirements.txt` and `/pyproject.toml`.

The Python package `pipreqs` can do this for `.py` files, but not for `.ipynb` files.

```python
# printing the contents of the shell script used
!cat ./search_script.sh
```

**<span style="color:gold">Note:</span>** `search_script.sh` assumes the project root directory is `pwd`'s "grandparent" (i.e., `./../..`); change if necessary.

```python
! ./search_script.sh
```

```python
import re
with open('./search_results.txt','r') as file:
    results = file.readlines()
```

```python
X = []

for s in results:
    # First grab `from ____` and ignore what follows
    try:
        m = re.search(r'from [\w\.\_]+',s)
        X.append(m.group(0))

    # What's left should be plain `import ____` lines
    except AttributeError:
        try:
            m = re.findall(r'import [\w_]+|,\s[\w]+', s)
            X.extend(m)
        except:
            print(f"failed on {s}")
            continue

X = [re.sub(r'from |import |, ','', s) for s in X]
X = list(set(X))
for s in X : print(s)
```
